# Exploratory analysis
This notebook inspects the synthetic Kaggle cohort. It does not establish clinical validity.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data/raw/alzheimers_disease_data.csv')
df.shape, df.head()
```

```python
summary = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'missing': df.isna().sum(),
    'unique': df.nunique(),
})
summary
```

```python
ax = sns.countplot(data=df, x='Diagnosis')
ax.set(title='Target distribution', xlabel='Diagnosis', ylabel='Patients')
plt.show()
```

```python
numeric = df.select_dtypes('number').drop(columns=['PatientID'])
correlations = numeric.corr()['Diagnosis'].drop('Diagnosis').sort_values(key=abs, ascending=False)
correlations.head(15)
```

## Leakage and interpretation notes
- `PatientID` is an identifier and is excluded from modeling.
- `DoctorInCharge` is constant/confidential metadata and is excluded.
- All preprocessing is fit inside model pipelines.
- Associations and feature importance values are not causal or clinically validated.
